```python
import pandas as pd

##LAD = Local Authority District: The main administrative divisions in the UK (like councils)
df0_bus = pd.read_parquet(raw_data / "business_counts" / "Business_counts_IS8_LADs.parquet", engine="fastparquet") 
df0_em = pd.read_parquet(raw_data / "employee_counts" / "Employee_counts_IS8_LADs.parquet", engine="fastparquet")
```

```python
group_keys = ["YEAR", "GEOGRAPHY_CODE", "GEOGRAPHY_NAME", "IS8_SECTOR", "FRONTIER_SECTOR"]

df0_bus.loc[df0_bus["IS8_SECTOR"] == "Digital and Technology", "FRONTIER_SECTOR"] = "Digital and Technology"
df0_bus.loc[df0_bus["IS8_SECTOR"] == "Defence sector", "FRONTIER_SECTOR"] = "Defence Sector"
df0_bus.loc[df0_bus["IS8_SECTOR"] == "Life Sciences", "FRONTIER_SECTOR"] = "Life Sciences"


df0_em.loc[df0_em["IS8_SECTOR"] == "Digital and Technology", "FRONTIER_SECTOR"] = "Digital and Technology"
df0_em.loc[df0_em["IS8_SECTOR"] == "Defence sector", "FRONTIER_SECTOR"] = "Defence Sector"
df0_em.loc[df0_em["IS8_SECTOR"] == "Life Sciences", "FRONTIER_SECTOR"] = "Life Sciences"


##This groupby is necessary because we have multiple columns in the businees data because is disaggregate by industry sector and size band. For us is interesting at the frontier
## level
df0_bus_grouped = df0_bus.groupby(group_keys, as_index=False)["OBS_VALUE"].sum()
df0_bus_grouped = df0_bus_grouped.rename(columns={"OBS_VALUE": "OBS_Value_Business"})
print(df0_bus_grouped['IS8_SECTOR'].unique())

## Same here, but the employment data is disaggregate by industry sector.
df0_em_grouped = df0_em.groupby(group_keys, as_index=False)["OBS_VALUE"].sum()
df0_em_grouped = df0_em_grouped.rename(columns={"OBS_VALUE": "OBS_Value_Employment"})
print(df0_em_grouped['IS8_SECTOR'].unique())

## Prioritize the business data available and  only we keep those values from employment data that math with business one.
df_macro_data= df0_bus_grouped.merge(
    df0_em_grouped,
    on=group_keys,
    how="inner",  ##First we have left, but I change this for inner to only keep those values that match in both datasets. Then we have a dataset from 2016 to 2024.      
    indicator=True    
)

print(df_macro_data["_merge"].value_counts())


```

```python
for col in group_keys:  
    df_macro_data[col] = df_macro_data[col].astype("category")

df_macro_data["OBS_Value_Business"] = df_macro_data["OBS_Value_Business"].astype("float32")
df_macro_data["OBS_Value_Employment"] = df_macro_data["OBS_Value_Employment"].astype("float32")
df_macro_data = df_macro_data.drop(columns=["_merge"])
df_macro_data.head()

df_macro_data.to_excel(clean_data / "Macro_data_business_employment.xlsx", sheet_name="data", index=False)
df_macro_data.to_csv(clean_data / "Macro_data_business_employment.csv", index=False)
```

```python
df1_bus = pd.read_parquet(raw_data / "business_counts" / "Business_counts_IS8_MSOAs.parquet", engine="fastparquet") 
df1_em = pd.read_parquet(raw_data / "employee_counts" / "Employee_counts_IS8_MSOAs.parquet", engine="fastparquet")
```

```python
print(df1_bus['IS8_SECTOR'].unique())
print(df1_em['IS8_SECTOR'].unique())
print(df1_bus['FRONTIER_SECTOR'].unique())
print(df1_em['FRONTIER_SECTOR'].unique())
```

```python
df1_bus.loc[df1_bus["IS8_SECTOR"] == "Digital and Technology", "FRONTIER_SECTOR"] = "Digital and Technology"
df1_bus.loc[df1_bus["IS8_SECTOR"] == "Defence sector", "FRONTIER_SECTOR"] = "Defence Sector"
df1_bus.loc[df1_bus["IS8_SECTOR"] == "Life Sciences", "FRONTIER_SECTOR"] = "Life Sciences"


df1_em.loc[df1_em["IS8_SECTOR"] == "Digital and Technology", "FRONTIER_SECTOR"] = "Digital and Technology"
df1_em.loc[df1_em["IS8_SECTOR"] == "Defence sector", "FRONTIER_SECTOR"] = "Defence Sector"
df1_em.loc[df1_em["IS8_SECTOR"] == "Life Sciences", "FRONTIER_SECTOR"] = "Life Sciences"


keys = ["YEAR","GEOGRAPHY_CODE", "GEOGRAPHY_NAME", "IS8_SECTOR", "FRONTIER_SECTOR"]

df1_bus_grouped = df1_bus.groupby(keys, as_index=False)['OBS_VALUE'].sum()
df1_bus_grouped = df1_bus_grouped.rename(columns={'OBS_VALUE': 'OBS_Value_Business'})

df1_em_grouped = df1_em.groupby(keys, as_index=False)['OBS_VALUE'].sum()
df1_em_grouped = df1_em_grouped.rename(columns={'OBS_VALUE': 'OBS_Value_Employment'})


df_micro_data = df1_bus_grouped.merge(df1_em_grouped, on=keys, how='inner', indicator=True)

print(df_micro_data["_merge"].value_counts())
     
```

```python
for col in keys:
    df_micro_data[col] = df_micro_data[col].astype('category')

df_micro_data['OBS_Value_Business']   = df_micro_data['OBS_Value_Business'].astype('float32')
df_micro_data['OBS_Value_Employment'] = df_micro_data['OBS_Value_Employment'].astype('float32')
df_micro_data = df_micro_data.drop(columns=["_merge"])

df_micro_data.head()
df_micro_data.to_excel(clean_data / "Micro_data_business_employment.xlsx", sheet_name="data", index=False)
df_micro_data.to_csv(clean_data / "Micro_data_business_employment.csv", index=False)
```

```python
#Combined Authority / City Region: Es una agrupación voluntaria de varios LADs que se unen para gobernar juntos temas como transporte, vivienda y desarrollo 
#económico. Los más famosos son:
    #Greater Manchester Combined Authority  → agrupa 10 LADs
    #West Midlands Combined Authority       → agrupa Birmingham + 6 LADs
    #West Yorkshire Combined Authority      → agrupa Leeds + 4 LADs

import pandas as pd
# Skip first 4 sheets
xls = pd.ExcelFile(raw_data / "local_indicators" / "ONS_local_indicators_package.xlsx")
print(xls.sheet_names)
sheet_names = xls.sheet_names[4:]
```

```python
dfs = {}

for sheet_name in sheet_names:
    raw_df = pd.read_excel(xls, sheet_name=sheet_name, header=None)
    
    area_row_idx = raw_df[raw_df.iloc[:,0].astype(str).str.contains("Area Code", na=False)].index
    if len(area_row_idx) == 0:
        print(f"Skipping sheet '{sheet_name}' — no 'Area Code' found")
        continue
    header_row = area_row_idx[0]
    
    df = pd.read_excel(xls, sheet_name=sheet_name, header=header_row)
    
    ci_cols = [c for c in df.columns if "Lower 95% Confidence Interval" in str(c) or "Upper 95% Confidence Interval" in str(c)]
    df = df.drop(columns=ci_cols)

    cols = []
    for c in df.columns:
        if c == 'Area Code':
            cols.append(c)
        else:
            cols.append(f"{c} [{sheet_name}]")
    df.columns = cols
    
    dfs[sheet_name] = df

df_local_indicators = None
for sheet_name, df in dfs.items():
    if df_local_indicators is None:
        df_local_indicators = df
    else:
        df_local_indicators = df_local_indicators.merge(df, on='Area Code', how='outer')

print(df_local_indicators.shape)
df_local_indicators.head()
```

```python
df_local_indicators.to_excel(clean_data / "local_indicators_all_sample.xlsx", sheet_name="_all_local_indicators", index=False)
df_local_indicators.to_csv(clean_data / "local_indicators_all_sample.csv", index=False)
```

```python
##Merge between macro data and local indicators both at the LAD level.
df_LAD_local_indicators = df_macro_data_2022.merge(
    df_local_indicators,
    left_on='GEOGRAPHY_CODE', 
    right_on='Area Code',   
    how='inner', 
    indicator=True    
)

print(df_LAD_local_indicators["_merge"].value_counts())

df_LAD_local_indicators.head()
```

```python
patterns = {
    'County or Unitary Authority': 'County_or_Unitary_Authority',
    'Region': 'Region',
    'Nation': 'Nation',
    'Country': 'Country'
}

for pattern, new_name in patterns.items():
    # Find all columns matching the pattern
    cols = [c for c in df_LAD_local_indicators.columns if c.startswith(pattern)]
    
    print(f"{pattern}: {cols}")  # verify before dropping
    
    # Keep first, drop the rest, rename
    df_LAD_local_indicators = df_LAD_local_indicators.drop(columns=cols[1:])
    df_LAD_local_indicators = df_LAD_local_indicators.rename(columns={cols[0]: new_name})
print(df_LAD_local_indicators.columns.tolist())
```
