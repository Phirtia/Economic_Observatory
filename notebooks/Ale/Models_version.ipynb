```python
#Combined Authority / City Region: Es una agrupación voluntaria de varios LADs que se unen para gobernar juntos temas como transporte, vivienda y desarrollo 
#económico. Los más famosos son:
    #Greater Manchester Combined Authority  → agrupa 10 LADs
    #West Midlands Combined Authority       → agrupa Birmingham + 6 LADs
    #West Yorkshire Combined Authority      → agrupa Leeds + 4 LADs

import pandas as pd
# Skip first 4 sheets
xls = pd.ExcelFile(raw_data / "local_indicators" / "ONS_local_indicators_package.xlsx")
print(xls.sheet_names)
sheet_names = xls.sheet_names[4:]
```

```python
dfs = []

for sheet_name in sheet_names:
    # Read the whole sheet first as raw
    raw_df = pd.read_excel(xls, sheet_name=sheet_name, header=None)
    
    # Find the row where 'Area Code' appears
    area_row_idx = raw_df[raw_df.iloc[:,0].astype(str).str.contains("Area Code", na=False)].index
    if len(area_row_idx) == 0:
        print(f"Skipping sheet '{sheet_name}' because no 'Area Code' found")
        continue
    header_row = area_row_idx[0]
    
    # Now read the sheet again using the correct header
    df = pd.read_excel(xls, sheet_name=sheet_name, header=header_row)
    
    # Drop confidence interval columns if they exist
    ci_cols = [c for c in df.columns if "Lower 95% Confidence Interval" in str(c) or "Upper 95% Confidence Interval" in str(c)]
    df = df.drop(columns=ci_cols)

    # Drop completely empty columns
    df = df.dropna(axis=1, how='all')
    
    # Identify numeric columns (including numbers stored as text)
    numeric_cols = []
    for col in df.columns:
        temp = pd.to_numeric(df[col], errors='coerce')
        if temp.notna().any():
            numeric_cols.append(col)
    
    if not numeric_cols:
        continue  # skip if no numeric column
    
    # Identifier columns
    id_cols = [c for c in df.columns if c not in numeric_cols]

    df_long = df.melt(
        id_vars=id_cols,
        value_vars=numeric_cols,
        var_name='Metric',
        value_name='Metric_Value'
    )

    df_long['Metric_Value'] = pd.to_numeric(df_long['Metric_Value'], errors='coerce')
    df_long['Sheet'] = sheet_name
    dfs.append(df_long)

df_all = pd.concat(dfs, ignore_index=True)

code_cols = [c for c in df_all.columns if 'Code' in c or 'Area' in c]
for col in code_cols:
    df_all[col] = df_all[col].astype('category')

print(df_all.info())
df_all.head(10)
```

```python
cols = [
    "Area Code",
    "Country",
    "Nation",
    "Region",
    "Combined Authority or City Region",
    "County or Unitary Authority",
    "Local Authority District",
    "Metric",
    "Metric_Value",
    "Notes",
    "Data accuracy",
    "Observation Status",
    "Sheet",
    "ITL Level 1",
    "ITL Level 2",
    "ITL Level 3",
    "Welsh Health Board",
    "Police Force Area"
]

df_all = df_all[cols]
##Long format is appropiate for graphs, but when we want to run regressions, maybe we need to come back to wide format?
df_all.to_excel(clean_data / "local_indicators_Diane.xlsx", sheet_name="_all_local_indicators", index=False)
```
